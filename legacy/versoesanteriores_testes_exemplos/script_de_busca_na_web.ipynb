#Duckduckgo

```python
from langchain_community.tools import DuckDuckGoSearchRun

def duckduckgo_search(query):
  search = DuckDuckGoSearchRun()
  return search.invoke(query)
```

```python
print(duckduckgo_search("Comer leite com manga faz mal?"))
```

```python
print(duckduckgo_search("Elon Musk é o homem mais rico do mundo?"))
```

```python
print(duckduckgo_search("Lula está preso"))
```

# Tavily

```python
import os
from google.colab import userdata
os.environ["TAVILY_API_KEY"] =userdata.get('TAVILY_API_KEY')
```

```python
from langchain_community.tools import TavilySearchResults

def tavily_search(query):
  tool = TavilySearchResults(
      max_results=3,
      search_depth="advanced",
      include_answer=True,
      include_raw_content=True,
      include_images=False
  )
  return tool.invoke({"query": query})
```

```python
response = tavily_search("Comer leite com manga faz mal?")
response
```

```python
response = tavily_search("Elon Musk é o homem mais rico do mundo?")
response
```

```python
response = tavily_search("Lula está preso")
response
```
